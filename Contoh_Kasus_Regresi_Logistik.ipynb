Berikut adalah tutorial menyelesaikan tugas regresi linear logistik

1. Import library python untuk AI

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Deskripsi:
*   Pandas : library python untuk manipulasi, pembersihan, dan analisis data
*   Scikit-learn (sklearn) : library python untuk machine learning

2. Masukkan data

```python
data = {
    'harga': [100, 120, 80, 150, 90, 110, 130],
    'rating': [4.8, 4.7, 4.5, 4.2, 4.9, 4.6, 4.3],
    'review': [200, 150, 300, 80, 400, 220, 100],
    'laris': [1, 1, 1, 0, 1, 1, 0]
}
df = pd.DataFrame(data)
print (df)
```

3. Menentukan variabel

```python
X = df[['harga', 'rating', 'review']]
y = df['laris']
```

Penjelasan :
*   `X = df[['harga', 'rating', 'review']]` : Nilai X itu diambil dari dataframe harga, rating, review yang sebagai fitur
*   `y = df['laris']` : Nilai y itu diambil dari dataframe laris yang sebagai target

4. Pembuatan X training, X testing dan y training, y testing dengan test 20% dari dataframe

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

5. Membuat model dari machine learning. Disini kita menggunakan model Regresi Logistik

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

Penjelasan

*  `model.fit(x_train, y_train)` adalah baris kode dalam Machine Learning (menggunakan Python, biasanya scikit-learn atau TensorFlow/Keras) yang berfungsi untuk melatih model menggunakan data latih

6. Membuat prediksi kedepan

```python
y_pred = model.predict(X_test)
print("Akurasi:", accuracy_score(y_test, y_pred))
print("Prediksi:", model.predict([[110, 4.8, 250]]))
```

7. Membuat visualisasi plot dari model linear Regresi ini

```python
plt.scatter(df['harga'], df['rating'], c=df['review'])
plt.xlabel("harga")
plt.ylabel("rating")
plt.title("Klasifikasi Laris")
plt.colorbar(label="Laris (0=tidak,1=ya)")
plt.show()
```

Hasil analisis praktikum

Pada regresi logistik, pendekatan yang digunakan berbeda karena output yang dihasilkan bukan angka kontinu melainkan kategori (laris atau tidak laris). Model ini digunakan untuk membantu pengambilan keputusan sebelum menjual produk. Berdasarkan hasil prediksi dan akurasi, model mampu mengklasifikasikan data dengan cukup baik meskipun masih terbatas pada data sederhana. Secara keseluruhan, ketiga metode ini memiliki peran yang berbeda: regresi linear untuk prediksi nilai, regresi berganda untuk analisis multi faktor, dan regresi logistik untuk klasifikasi keputusan dalam bisnis.
